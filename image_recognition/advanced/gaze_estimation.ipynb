# 発展：視線推定（L2CS-Net）

Webカメラ映像から視線の方向を推定します．**PyTorch と L2CS-Net が必要**で，標準の `pbl2026` 環境には含めていません．
視線を使うプロジェクトを行うグループだけ，次の手順で追加してください．

```bash
conda activate pbl2026
pip install torch torchvision --index-url https://download.pytorch.org/whl/cpu
pip install git+https://github.com/edavalosanaya/L2CS-Net.git@main
```

学習済みモデル `L2CSNet_gaze360.pkl` は [L2CS-Net 公式リポジトリ](https://github.com/Ahmednull/L2CS-Net) の README にある Google Drive のリンク（Gaze360）からダウンロードし，`det/` に置きます．

> 昨年度までの「視線追跡システム」プロジェクトで使用した資料です．

````python
from l2cs import Pipeline, render
import cv2
import torch

gaze_pipeline = Pipeline(
    weights='det/L2CSNet_gaze360.pkl',
    arch='ResNet50',
    device=torch.device('cpu') # or 'gpu'
)
 
cap = cv2.VideoCapture(0)

with torch.no_grad():
    while True:
        ret, frame = cap.read()
        if not ret:
            print("フレームの取得に失敗しました")
            break
    
        # フレームを処理して可視化
        results = gaze_pipeline.step(frame)
        frame = render(frame, results)
    
        # 画像を表示
        cv2.imshow('Gaze Estimation', frame)
    
        # 'q'キーが押されたらループを終了
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

# リソースの解放
cap.release()
cv2.destroyAllWindows()
````
